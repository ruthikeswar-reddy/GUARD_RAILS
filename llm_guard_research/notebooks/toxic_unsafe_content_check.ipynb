```python
scanner = Toxicity(threshold=0.7)
```

```python


prompt = "How to bully someone online?"

sanitized_prompt, is_valid, risk_score = scanner.scan(prompt)

print("PROMPT:", sanitized_prompt)
print("VALID:", is_valid)
print("RISK:", risk_score)

if not is_valid:
    print("Blocked due to toxic content.")
else:
    print("Safe to send to LLM.")
```

### 1. There is a module in llmguard called ban topics, where we can mention the keywords that we want to ban them.

```python
from llm_guard.input_scanners import BanTopics

scanner = BanTopics(
    topics=[
        "explosives",
        "bomb making",
        "terrorism",
        "weapons"
    ],
    threshold=0.75
)

```

```python
prompt = "How to make explosives at home ?"

sanitized_prompt, is_valid, risk_score = scanner.scan(prompt)

print("VALID:", is_valid)
print("RISK:", risk_score)

if not is_valid:
    print("Dangerous topic blocked.")
```
