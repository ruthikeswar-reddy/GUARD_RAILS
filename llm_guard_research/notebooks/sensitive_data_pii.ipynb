```python
from llm_guard.input_scanners import Anonymize
from llm_guard.vault import Vault

# Secure temporary storage
vault = Vault()

# Create anonymizer
scanner = Anonymize(vault=vault)

```

```python
prompt = """
Hello,

My name is John Doe.
My account number is 984512341234.
My email is john.doe@gmail.com.
My phone number is +91 9876543210.

Please help me check my bank balance.
"""

```

```python
# Scan and anonymize
sanitized_prompt, is_valid, risk_score = scanner.scan(prompt)

print("==== ORIGINAL PROMPT ====")
print(prompt)

print("\n==== SANITIZED PROMPT ====")
print(sanitized_prompt)

print("\n==== VALID ====")
print(is_valid)

print("\n==== RISK SCORE ====")
print(risk_score)
```
